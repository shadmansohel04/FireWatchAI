```python
import os
import shutil
from ultralytics import YOLO

source_dir = "/kaggle/input/smoke-fire-detection-yolo/data"
dest_dir = "/content/dataset"

# 1. Copy dataset to writable Colab storage (takes ~1-2 min)
if not os.path.exists(dest_dir):
    print("Copying dataset to writable disk...")
    shutil.copytree(source_dir, dest_dir)
    print("Done copying.")

# 2. Write data.yaml targeting the writable path
yaml_path = "/content/data.yaml"
yaml_content = f"""
path: {dest_dir}
train: train/images
val: val/images
test: test/images

nc: 2
names: ['smoke', 'fire']
"""

with open(yaml_path, "w") as f:
    f.write(yaml_content.strip())

# 3. Start training with full access
model = YOLO("yolo11n.pt")
results = model.train(
    data=yaml_path,
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    patience=10,
    mosaic=0.5,
    save=True
)
```
